```python
import json
import numpy as np

import pandas as pd

from tqdm import tqdm

from sentence_transformers import SentenceTransformer

from sklearn.metrics.pairwise import cosine_similarity

import matplotlib.pyplot as plt

from collections import Counter
```

```python
INPUT_FILE="/content/drive/MyDrive/fake_news_dataset_30k.json"

with open(INPUT_FILE,"r",encoding="utf8") as f:
    data=json.load(f)

articles=data["articles"]

print(len(articles))
```

```python
texts=[]

for article in articles:

    txt=article["title"]+" "+article["text"]

    texts.append(txt)
```

```python
import torch

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

model = SentenceTransformer(
    "BAAI/bge-m3",
    device=device
)
```

```python
embeddings=model.encode(
    texts,
    batch_size=16,
    show_progress_bar=True,
    normalize_embeddings=True
)
```

```python
print(type(embeddings))

print(embeddings.shape)

print()

print("Nombre d'articles :", len(texts))

print("Dimension embedding :", embeddings.shape[1])
```

## Pipeline Final : Extraction du Dataset Représentatif (10k)
Ce pipeline automatise la réduction de dimension et le clustering pour passer de 30k à 10k articles tout en préservant la diversité sémantique.

```python
import time
import json
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import pairwise_distances_argmin_min

# --- CONFIGURATION ---
N_CLUSTERS = 10000
PCA_COMPONENTS = 128
OUTPUT_FILE = "fake_news_dataset_10k_final.json"

print(f"--- Démarrage du Pipeline d'Optimisation (KMeans Haute Performance) ---")
start_total = time.time()

# 1. Réduction de dimension (PCA)
print(f"1/4. Réduction PCA ({embeddings.shape[1]} -> {PCA_COMPONENTS})...")
pca = PCA(n_components=PCA_COMPONENTS, random_state=42)
embeddings_reduced = pca.fit_transform(embeddings)
print(f"   > Variance conservée : {pca.explained_variance_ratio_.sum()*100:.2f}%")

# 2. Clustering KMeans Ultra-Accéléré
# algorithm='elkan' est plus rapide, tol=1e-3 et max_iter=50 pour accélérer la convergence
print(f"2/4. Clustering KMeans ({N_CLUSTERS} clusters, convergence rapide)...")
kmeans = KMeans(
    n_clusters=N_CLUSTERS,
    random_state=42,
    n_init=1,
    algorithm='elkan',
    max_iter=50,
    tol=1e-3,
    verbose=0
)

kmeans.fit(embeddings_reduced)

# 3. Sélection des articles représentatifs
print("3/4. Sélection des articles les plus proches des centroïdes...")
closest_indices, _ = pairwise_distances_argmin_min(kmeans.cluster_centers_, embeddings_reduced)
representative_indices = sorted(list(set(closest_indices)))
new_articles_10k = [articles[i] for i in representative_indices]
print(f"   > Articles uniques sélectionnés : {len(new_articles_10k)}")

# 4. Sauvegarde
with open(OUTPUT_FILE, "w", encoding="utf-8") as f:
    json.dump({"articles": new_articles_10k}, f, ensure_ascii=False, indent=4)

print(f"4/4. Succès ! Dataset sauvegardé dans {OUTPUT_FILE}")
print(f"--- Temps total : {time.time() - start_total:.2f}s ---")

# --- VISUALISATION ---
def plot_final_dist(orig, red):
    c1, c2 = Counter(a['category'] for a in orig), Counter(a['category'] for a in red)
    cats = sorted(c1.keys())
    x = np.arange(len(cats))
    plt.figure(figsize=(12, 5))
    plt.bar(x - 0.2, [c1[c] for c in cats], 0.4, label='Initial (30k)', color='#bdc3c7')
    plt.bar(x + 0.2, [c2[c] for c in cats], 0.4, label='Final (10k)', color='#1abc9c')
    plt.xticks(x, cats, rotation=45)
    plt.title("Représentativité Sémantique par Catégorie")
    plt.legend()
    plt.grid(axis='y', alpha=0.3)
    plt.show()

plot_final_dist(articles, new_articles_10k)
```

```python
cluster_sizes = Counter(kmeans.labels_)

print()

print("Nombre de clusters :", len(cluster_sizes))

print("Cluster min :", min(cluster_sizes.values()))

print("Cluster max :", max(cluster_sizes.values()))

print("Cluster moyen :", np.mean(list(cluster_sizes.values())))
```

```python
# --- VISUALISATION DE LA QUALITÉ ---
def plot_comparison(orig, reduced):
    c_orig = Counter(a['category'] for a in orig)
    c_red = Counter(a['category'] for a in reduced)
    cats = sorted(c_orig.keys())

    x = np.arange(len(cats))
    plt.figure(figsize=(12, 6))
    plt.bar(x - 0.2, [c_orig[c] for c in cats], 0.4, label='Original (30k)', color='lightgray')
    plt.bar(x + 0.2, [c_red[c] for c in cats], 0.4, label='Réduit (10k)', color='teal')
    plt.xticks(x, cats, rotation=45)
    plt.title("Maintien de la représentativité par catégorie")
    plt.legend()
    plt.tight_layout()
    plt.show()

plot_comparison(articles, new_articles_10k)
```
